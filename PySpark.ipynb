# **Crear Session para datos Fifa**

```python
import pyspark
from pyspark.sql import SparkSession, SQLContext
from pyspark.sql.types import IntegerType,StringType,StructType,StructField
from pyspark.sql.functions import col,lit
SparkSession = SparkSession\
              .builder\
              .appName("ETL_Basica")\
              .getOrCreate()
SQLContext=SQLContext(SparkSession)
```

# **Cargo dataset**

```python
players_23=SparkSession.read.csv("FIFA23_official_data.csv",header="true",sep=",")
players_23.show(5)
display(players_23)
```

```python
players_23=players_23.select('name','Position','Height','Weight','Release Clause','Work Rate','Skill Moves','Weak Foot') # defino las columnas que me interesan
```

```python
rdd=players_23.rdd # cargo el rdd con las columnas seleccionadas anteriormente
rdd.collect()
```

```python
schema=StructType([
    StructField('name',StringType(),False),
    StructField('Position',StringType(),True),
    StructField('Height',StringType(),True),
    StructField('Weight',StringType(),True),
    StructField('Release Clause',StringType(),True),
    StructField('Work Rate',StringType(),True),
    StructField('Skill Moves',StringType(),True),
    StructField('Weak Foot',StringType(),True)

])
```

```python
dataframe=SQLContext.createDataFrame(rdd,schema)
dataframe.printSchema()
```

```python
dataframe.show(5)
```

# **Elimino valores nulos del dataframe**

```python
dataframe.count()
```

```python
dataframe=dataframe.dropna()
dataframe.count()
```

# **# Convierto columas de String a integer**

```python
dataframe=dataframe.withColumn("Height",dataframe["Height"].cast(IntegerType()))
dataframe=dataframe.withColumn("Weight",dataframe["Weight"].cast(IntegerType()))
dataframe=dataframe.withColumn("Skill Moves",dataframe["Skill Moves"].cast(IntegerType()))
dataframe=dataframe.withColumn("Weak Foot",dataframe["Weak Foot"].cast(IntegerType()))
dataframe.printSchema()
dataframe.show(10)
```

# **calculo el avg entre skill moves and weak foot**

```python
dataframe=dataframe.withColumn('avg',((col('skill moves')+col('weak foot'))/lit(2)))
dataframe.show()
```

# **saco el output**

```python
dataframe.repartition(1).write.csv("output.csv",header=True,sep=",")
```
